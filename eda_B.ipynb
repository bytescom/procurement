Outlier Analysis

```python
# Inspect numerical distributions at different percentiles
dfb1[['quantity', 'line_amount', 'currency_rate']].quantile(
    [0, 0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99, 1]
)
```

```python
# Calculate IQR outlier counts for numerical columns
outlier_summary = {}
for col in ['quantity', 'line_amount', 'currency_rate']:
    x = dfb1[col].dropna()
    q1, q3 = x.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outlier_summary[col] = {
        'lower_bound': lower,
        'upper_bound': upper,
        'outlier_count': ((x < lower) | (x > upper)).sum()
    }
pd.DataFrame(outlier_summary).T
```

```python
# Calculate IQR outlier counts separately for each ERP system
def iqr_count(x):
    x = x.dropna()
    q1, q3 = x.quantile([0.25, 0.75])
    iqr = q3 - q1
    return ((x < q1 - 1.5 * iqr) | (x > q3 + 1.5 * iqr)).sum()

dfb1.groupby('source_system')[
    ['quantity', 'line_amount', 'currency_rate']
].agg(iqr_count)
```

**Note:** IQR is an investigation technique, not an automatic deletion rule. Different ERP systems may have different distributions.
