Outlier Investigation (IQR Method)

```python
# Outlier analysis
outlier_summary = []
for col in ['gross_amount', 'quantity']:
    q1 = dfA2[col].quantile(0.25)
    q3 = dfA2[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = dfA2[(dfA2[col] < lower_bound) | (dfA2[col] > upper_bound)]
    outlier_summary.append({
        'column': col,
        'q1': q1,
        'q3': q3,
        'iqr': iqr,
        'lower_bound': lower_bound,
        'upper_bound': upper_bound,
        'outlier_count': len(outliers),
        'outlier_pct': round(len(outliers) / len(dfA2) * 100, 2)
    })

pd.DataFrame(outlier_summary)
```
